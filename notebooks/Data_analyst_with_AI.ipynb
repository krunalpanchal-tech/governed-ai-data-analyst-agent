```python
def verified_sales_answer():
    total_sales = df["Sales"].sum()
    total_profit = df["Profit"].sum()

    return {
        "total_sales": round(float(total_sales), 2),
        "total_profit": round(float(total_profit), 2),
        "source": "Calculated directly from the loaded dataset"
    }

verified_sales_answer()
```

```python
def verify_sales():
    actual_sales = round(float(df["Sales"].sum()), 2)

    result = {
        "verified_sales": actual_sales,
        "verification_status": "Verified",
        "data_source": "Superstore dataset"
    }

    return result

verify_sales()
```

```python
def verify_business_metrics():
    actual_sales = round(float(df["Sales"].sum()), 2)
    actual_profit = round(float(df["Profit"].sum()), 2)

    return {
        "total_sales": actual_sales,
        "total_profit": actual_profit,
        "sales_verified": actual_sales == round(float(df["Sales"].sum()), 2),
        "profit_verified": actual_profit == round(float(df["Profit"].sum()), 2),
        "source": "Superstore dataset"
    }

verify_business_metrics()
```

```python
def verified_sales_response():
    metrics = verify_business_metrics()

    return (
        f"Total Sales: ${metrics['total_sales']:,.2f}\n"
        f"Total Profit: ${metrics['total_profit']:,.2f}\n"
        f"Source: {metrics['source']}\n"
        f"Verification: "
        f"{'Passed' if metrics['sales_verified'] and metrics['profit_verified'] else 'Failed'}"
    )

print(verified_sales_response())
```

```python
def verified_category_analysis():
    result = (
        df.groupby("Category")
        .agg(
            Total_Sales=("Sales", "sum"),
            Total_Profit=("Profit", "sum")
        )
        .round(2)
        .sort_values("Total_Profit", ascending=False)
    )

    result["Profit_Margin_%"] = (
        result["Total_Profit"] / result["Total_Sales"] * 100
    ).round(2)

    return result

verified_category_analysis()
```

```python
def ask_category_analyst():
    category_data = verified_category_analysis().to_string()

    prompt = f"""
    You are a business data analyst.

    Use only the verified category data below.
    Explain:
    1. Which category has the highest profit.
    2. Which category has the highest sales.
    3. Which category has the highest profit margin.
    4. One practical business recommendation.

    Do not change, estimate, or invent any numbers.
    If a value is not available, say so.

    Verified category data:
    {category_data}
    """

    response = ollama.chat(
        model="llama3.2:3b",
        messages=[
            {
                "role": "system",
                "content": "Explain business data clearly. Treat supplied data as the only source of facts."
            },
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    return response["message"]["content"]


print(ask_category_analyst())
```

```python
def governed_category_analyst(question):
    allowed_keywords = [
        "category", "sales", "profit",
        "margin", "performance"
    ]

    if not any(word in question.lower() for word in allowed_keywords):
        return "Blocked: This question is outside the approved category-analysis scope."

    return ask_category_analyst()


print(governed_category_analyst(
    "Which category has the highest profit?"
))

print(governed_category_analyst(
    "Tell me a joke."
))
```

```python
def validate_category_totals():
    category_data = verified_category_analysis()

    category_sales = round(
        float(category_data["Total_Sales"].sum()), 2
    )
    category_profit = round(
        float(category_data["Total_Profit"].sum()), 2
    )

    overall_sales = round(float(df["Sales"].sum()), 2)
    overall_profit = round(float(df["Profit"].sum()), 2)

    return {
        "category_sales_match": abs(category_sales - overall_sales) < 0.02,
        "category_profit_match": abs(category_profit - overall_profit) < 0.02,
        "overall_sales": overall_sales,
        "overall_profit": overall_profit
    }

validate_category_totals()
```

```python
def project_health_check():
    metrics = verify_business_metrics()
    reconciliation = validate_category_totals()

    checks = {
        "Sales calculation": metrics["sales_verified"],
        "Profit calculation": metrics["profit_verified"],
        "Category sales reconciliation": reconciliation["category_sales_match"],
        "Category profit reconciliation": reconciliation["category_profit_match"]
    }

    for check, passed in checks.items():
        status = "PASS" if passed else "FAIL"
        print(f"{check}: {status}")

    print("\nOverall status:",
          "PASS" if all(checks.values()) else "FAIL")

project_health_check()
```

```python
def verified_monthly_analysis():
    data = df.copy()

    data["Order Date"] = pd.to_datetime(data["Order Date"])

    result = (
        data.groupby(data["Order Date"].dt.to_period("M"))
        .agg(
            Total_Sales=("Sales", "sum"),
            Total_Profit=("Profit", "sum"),
            Order_Count=("Order ID", "nunique")
        )
        .round(2)
    )

    result.index = result.index.astype(str)

    return result

verified_monthly_analysis()
```

```python
# Page 2 - Profitability Validation

# 1. Overall profit margin
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()

profit_margin = (total_profit / total_sales) * 100

# 2. Count loss-making products
product_profit = (
    df.groupby("Product Name")["Profit"]
    .sum()
)

loss_making_products = (product_profit < 0).sum()

# 3. Category profitability
category_profit = (
    df.groupby("Category")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum")
    )
)

category_profit["Profit_Margin_%"] = (
    category_profit["Total_Profit"] /
    category_profit["Total_Sales"] * 100
)

print("PROFITABILITY VALIDATION")
print("-" * 40)
print("Total Profit:", round(total_profit, 2))
print("Profit Margin:", round(profit_margin, 2), "%")
print("Loss-Making Products:", loss_making_products)

print("\nCATEGORY PROFITABILITY")
display(category_profit.round(2))
```

```python
# Page 3 - Customer & Regional Validation

total_customers = df["Customer ID"].nunique()
total_orders = df["Order ID"].nunique()
total_sales = df["Sales"].sum()

average_order_value = total_sales / total_orders

# Regional performance
region_analysis = df.groupby("Region").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Orders=("Order ID", "nunique")
).round(2)

# Top 10 customers
top_10_customers = (
    df.groupby("Customer Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .round(2)
)

print("CUSTOMER & REGIONAL VALIDATION")
print("-" * 40)

print("Total Customers:", total_customers)
print("Total Orders:", total_orders)
print("Average Order Value:", round(average_order_value, 2))
print("Total Sales:", round(total_sales, 2))

print("\nREGIONAL PERFORMANCE")
display(region_analysis)

print("\nTOP 10 CUSTOMERS")
display(top_10_customers)
```
